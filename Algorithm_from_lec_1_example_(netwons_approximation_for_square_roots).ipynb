```python
import math as m

a = int(input("a number whose square root you wanna find but dont know: "))
b = int(input("an intuitive guess of what that number could be...: "))

def thep(number, guess):
    anum = float(guess)

    print("guessed approximations")

    if round(anum*anum) != number:
        for _ in range(1000):
            anum = (anum + (number/anum))/2
            print(anum)
            if abs(anum*anum - number) < 1e-9:
              if abs(anum-round(anum)) < 1e-9:
                  return round(anum)
              return anum
    else:
        return guess

thep(a, b)
```
